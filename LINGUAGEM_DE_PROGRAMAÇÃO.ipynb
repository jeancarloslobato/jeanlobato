### Definição Completa da Classe `Pessoa` e Instanciação do Objeto

```python
class Pessoa:
  def __init__(self, nome, idade, genero):
    self.nome = nome
    self.idade = idade
    self.genero = genero

  def falar_nome(self):
    return f"Olá, meu nome é {self.nome}"

  def aniversario(self):
    self.idade += 1
    print(f"Parabéns! {self.nome} completou {self.idade} anos.")

# Instanciando o objeto (agora a classe já contém o método aniversario)
pessoa1 = Pessoa("jean", 23, "masculino")

# Exibindo o nome
print(pessoa1.falar_nome())

# Chamando o método aniversario
pessoa1.aniversario()

# Verificando a nova idade
print(f"Nova idade da pessoa1: {pessoa1.idade}")
```

```python
#3 Herança - classe-pai Animal e duas classes-filhas
class Animal:
  def __init__(self, nome, idade):
    self.nome = nome
    self.idade = idade

def fazer_barulho(self):
  pass #método "vazio" quem será sobrescrito pelas filhas
```

```python
class cachorro(Animal):
  def fazer_barulho(self):
    return "latir"
```

```python
class Gato(Animal):
  def fazer_barulho(self):
    return "Miar"
```

```python
# 4. Criando e usando objetos das classes.filhas
rex = cachorro("Rex", 3)
Whisker = Gato("Whisker", 2)
```

```python
print(f"{rex.nome} faz: {rex.fazer_barulho()}")
print(f"{Whisker.nome} faz: {Whisker.fazer_barulho()}")
```
