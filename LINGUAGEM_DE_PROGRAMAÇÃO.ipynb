```python
filmes = {"Filme 1", "Filme 2", "Filme 3", "Filme 4", "FIlme 5"}
print("Bem vindo a classificação do filme")
print("Você tem cinco filmes para classificar")
print("digite '0' a qualquer momento para parar.\n")

for filme_item in filmes:
  classificacao_str = input(f" como você classifica '{filme_item}' de 1 a 5 (ou 0 para parar):")

  if classificacao_str == '0':
    print ("Que pena")
    break

  try:
    classificacao = int(classificacao_str)
  except ValueError:
    print("Classificação inválida. Por favor, digite um número.")
    continue

  if classificacao < 1 or classificacao > 5:
    print (f"Classificação inválida")
  else:
    print(f"Você classificou '{filme_item}' com a '{classificacao}' estrelas.\n")

print("Obrigado por classificar os filmes")
```
